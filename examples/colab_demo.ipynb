# colab-rich 簡單示範

這個 notebook 示範如何在 Google Colab 使用簡單的 Rich 輸出工具。

```python
from colab_rich import (
    bullet_list, code, columns, error, info, markdown, panel,
    show_json, show_progress, success, table, title, warning
)

title('我的第一個 Colab 程式')
info('今天學習如何顯示漂亮的輸出')

table(
    ['姓名', '分數'],
    [
        ['小明', 90],
        ['小華', 85],
    ],
)

success('表格完成！')
warning('這只是示範訊息。')
error('這也是示範訊息，不是真正的錯誤。')

markdown('**Markdown** 也可以顯示。')
panel('這是一個重點提示。', '小提示')
code("print('你好！')")
bullet_list(['第一項', '第二項'])
columns(['紅色', '綠色', '藍色'])
show_json({'分數': 90, '合格': True})
```

```python
for number in range(1, 6):
    show_progress(number, 5, '處理中')
```

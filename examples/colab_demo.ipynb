```python
from colab_rich import error, info, show_progress, success, table, title, warning

title('我的第一個 Colab 程式')
info('今天學習如何顯示漂亮的輸出')

table(
    ['姓名', '分數'],
    [
        ['小明', 90],
        ['小華', 85],
    ],
)

success('表格完成！')
warning('這只是示範訊息。')
error('這也是示範訊息，不是真正的錯誤。')
```

```python
for number in range(1, 6):
    show_progress(number, 5, '處理中')
```
